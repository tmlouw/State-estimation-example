
# Particle Filter Animation: Leaching Reactor

*Companion to* `leaching_ekf.ipynb` *— same model, same scenario, illustrated with
a particle filter instead of the EKF.*

This notebook builds, for a single user-chosen time step `step_idx`, an animated
6-step picture of what a particle filter actually does under the hood:

1. Sampling — draw particles from the current belief.
2. Propagate — push each particle through the (stochastic) dynamics.
3. Observation model — map each propagated particle through $h(x)$.
4. Measure — compare predictions to the real measurement; weight particles by likelihood.
5. Update — form a new belief (here: fit a Gaussian to the weighted particles —
   a Gaussian particle filter, closely related to an ensemble Kalman filter, not
   a traditional particle filter) and let the old particles fade.
6. Repeat — sample fresh particles from the new belief.


```python
import os
import shutil
import subprocess
import numpy as np
import jax.numpy as jnp
from jax import vmap
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import Video, display

np.random.seed(0)
```


## Model

*Reused from* `leaching_ekf.ipynb`

$$
\frac{dC_{\textrm{Cu}}^{(s)}}{dt}  =  \frac{F}{V}\big(C_{\textrm{Cu},in}^{(s)} - C_{\textrm{Cu}}^{(s)}\big) -r \\[8pt]
\frac{dC_{\textrm{Cu}}^{(aq)}}{dt} = -\frac{F}{V} C_{\textrm{Cu}}^{(aq)} + r \\[8pt]
\frac{dC_{\textrm{H}}}{dt}         =  \frac{F}{V}\big(C_{\textrm{H},in} - C_{\textrm{H}}\big) - \gamma\, r + \frac{q_H}{V}  \\[12pt]
r = k\, C_{\textrm{H}}\, C_{\textrm{Cu}}^{(s)} \\[12pt]
pH = -\log_{10}\!\big(C_{\textrm{H}}\big) + v, \qquad v \sim \mathcal{N}(0, R)
$$

The relevant component concentrations are:
- Solid-phase (leachable) copper concentration: $C_{\textrm{Cu}}^{(s)}$
- Dissolved copper concentration: $C_{\textrm{Cu}}^{(aq)}$
- Acid (proton) concentration: $C_{\textrm{H}}$

The slurry stream has a constant volumetric flow $F$ in and out, carrying ore
(solid copper) and background acid. The acid dosing stream adds acid at a
molar flowrate $q_H$. We assume the volume of acid dosed is small so it
doesn't affect the tank volume $V$, only the acid mass balance. The
stoichiometric coefficient for acid consumption per unit Cu dissolved is
$\gamma$, and $C_{\textrm{H},in}$ is the acid entering with the slurry feed,
and $C_{\textrm{Cu},in}^{(s)}$ the incoming ore's copper content (the feed
grade).


```python
params = dict(
    F=1.0,           # slurry volumetric flow, in = out [m^3/min]
    V=10.0,          # reactor volume [m^3]
    k=0.05,          # reaction rate constant [1/(min * (mol/m^3))]
    gamma=1.0,       # stoichiometric acid consumption per unit Cu dissolved
    C_H_in=0.05,     # acid concentration in slurry feed [mol/m^3]
    C_Cu_s_in=8.0,   # solid-phase Cu concentration in feed [mol/m^3]
)
p = tuple(params.values())


def f(x, u, p):
    C_Cu, C_H, C_Cu_s = x
    q_H = u
    F, V, k, gamma, C_H_in, C_Cu_s_in = p

    r = k * C_H * C_Cu_s

    dC_Cu = -(F / V) * C_Cu + r
    dC_H = (F / V) * (C_H_in - C_H) - gamma * r + (q_H / V)
    dC_Cu_s = (F / V) * (C_Cu_s_in - C_Cu_s) - r

    return jnp.array([dC_Cu, dC_H, dC_Cu_s])


def h(x):
    C_Cu, C_H, C_Cu_s = x
    return jnp.array([-jnp.log10(C_H)])


def u_profile(t):
    return 2.5 if t < 60.0 else 1.25   # [mol/min], step at t = 60 min


# vmapped versions: evaluate f and h for a whole (N, 3) batch of particles at once
f_vec = vmap(f, in_axes=(0, None, None))
h_vec = vmap(h, in_axes=0)
```

Simulate the true plant and generate noisy pH measurements

```python
def plant_ode(t, x, p):
    return np.array(f(jnp.array(x), u_profile(t), p))


T_final = 180.0
t_grid = np.linspace(0.0, T_final, 2000)
t_meas = np.arange(0.0, T_final + 2.0, 2.0)

R = np.array([[0.05]])                     # pH measurement noise variance
Q_proc = np.diag([1e-2, 1e-2, 1e-2])       # process noise covariance (continuous-time)

x0_true = np.array([2.0, 0.6, 6.0])
sol = solve_ivp(plant_ode, (t_grid[0], t_grid[-1]), x0_true, t_eval=t_grid, args=(p,))
x_true = sol.y.T

idx_meas = np.searchsorted(t_grid, t_meas)
sigma_meas = np.sqrt(R[0, 0])
y_true_full = np.array([h(jnp.array(x_true[i]))[0] for i in range(len(t_grid))])
y_meas = np.array([y_true_full[idx] for idx in idx_meas]) + sigma_meas * np.random.randn(len(t_meas))
```

## Particle filter

```python
def rk4_step_vec(X, u, p, dt):
    """RK4 step for a whole (N,3) batch of particles at once (via vmap)."""
    k1 = f_vec(X, u, p)
    k2 = f_vec(X + 0.5 * dt * k1, u, p)
    k3 = f_vec(X + 0.5 * dt * k2, u, p)
    k4 = f_vec(X + dt * k3, u, p)
    return X + (dt / 6.0) * (k1 + 2 * k2 + 2 * k3 + k4)


def propagate_particles(X0, u, p, Qc, dt, n_sub, rng):
    """Propagate all particles from t to t+dt, returning the FULL trajectory
    (n_sub+1, N, 3) so we can animate the path, not just the endpoint."""
    N, n = X0.shape
    traj = np.zeros((n_sub + 1, N, n))
    traj[0] = X0
    dt_sub = dt / n_sub
    L = np.linalg.cholesky(Qc + 1e-12 * np.eye(n))

    X = jnp.array(X0)
    for s in range(1, n_sub + 1):
        X = rk4_step_vec(X, u, p, dt_sub)
        noise = (rng.standard_normal((N, n)) @ L.T) * np.sqrt(dt_sub)
        X = jnp.array(np.maximum(np.array(X) + noise, 1e-8))
        traj[s] = np.array(X)
    return traj


def compute_weights(y_pred, y_meas_k, Rm):
    sigma2 = Rm[0, 0]
    logw = -0.5 * (y_meas_k - y_pred) ** 2 / sigma2
    logw -= logw.max()          # normalize for numerical stability
    w = np.exp(logw)
    return w / w.sum()


def weighted_mean_cov(X, w):
    mean = np.average(X, axis=0, weights=w)
    Xc = X - mean
    cov = (Xc.T * w) @ Xc + 1e-8 * np.eye(X.shape[1])
    return mean, cov


def run_particle_filter(x0_hat, P0, t_meas, y_meas, p, Qc, Rm, N=100, n_sub=20, seed=0):
    """Runs the Gaussian particle filter across all measurement times, storing
    everything needed to regenerate the 5-step figure set for ANY chosen
    transition afterward, without re-running the filter."""
    rng = np.random.default_rng(seed)
    x_hat, P = x0_hat.copy(), P0.copy()
    history = []

    for k in range(1, len(t_meas)):
        t_prev, t_next = t_meas[k - 1], t_meas[k]
        dt = t_next - t_prev
        u = u_profile(t_prev)

        # 1. Sample
        X0 = np.maximum(rng.multivariate_normal(x_hat, P, size=N), 1e-8)

        # 2. Propagate
        traj = propagate_particles(X0, u, p, Qc, dt, n_sub, rng)
        Xk = traj[-1]

        # 3. Observation model
        y_pred = np.array(h_vec(jnp.array(Xk)))[:, 0]

        # 4. Measure -> weights
        w = compute_weights(y_pred, y_meas[k], Rm)

        # 5. Update -> new Gaussian belief
        x_hat_new, P_new = weighted_mean_cov(Xk, w)

        history.append(dict(
            t_prev=t_prev, t_next=t_next, dt=dt,
            x_hat_prior=x_hat.copy(), P_prior=P.copy(),
            X0=X0, traj=traj, Xk=Xk,
            y_pred=y_pred, y_meas=y_meas[k], w=w,
            x_hat_post=x_hat_new.copy(), P_post=P_new.copy(),
        ))

        x_hat, P = x_hat_new, P_new   # 6. Repeat

    return history
```

Run the particle filter over the entire simulation time

```python
x0_hat = np.array([0.0, 2.0, 1.0])          # same deliberately wrong guess as the EKF
P0 = np.diag([1.0, 0.1, 4.0])

history = run_particle_filter(x0_hat, P0, t_meas, y_meas, p, Q_proc, R, N=100, n_sub=20, seed=0)
print(f"Particle filter ran for {len(history)} steps.")
```


## Visualization

A 2x2 panel layout: dissolved Cu, acid, solid Cu (top row and bottom-left),
and pH (bottom-right). Particle markers are dark blue throughout (including
in measurement space); the true measurement is dark yellow; belief summaries
(95% CI) are black error bars, offset slightly left of the markers so they
don't overlap.


```python
def compute_axis_limits(steps, R, pad_frac=0.08):
    state_data = [[] for _ in range(3)]
    y_meas_window = np.array([s["y_meas"] for s in steps])
    ph_data = [y_meas_window]

    for s in steps:
        for i in range(3):
            state_data[i].append(s["X0"][:, i])
            state_data[i].append(s["Xk"][:, i])
            std_prior = np.sqrt(s["P_prior"][i, i])
            std_post = np.sqrt(s["P_post"][i, i])
            state_data[i].append(np.array([
                s["x_hat_prior"][i] - 1.96 * std_prior, s["x_hat_prior"][i] + 1.96 * std_prior,
                s["x_hat_post"][i] - 1.96 * std_post, s["x_hat_post"][i] + 1.96 * std_post,
            ]))
        ph_data.append(s["y_pred"])

    sigma = np.sqrt(R[0, 0])
    ph_data.append(y_meas_window - 1.96 * sigma)
    ph_data.append(y_meas_window + 1.96 * sigma)

    state_ylim = []
    for i in range(3):
        arr = np.concatenate(state_data[i])
        lo, hi = arr.min(), arr.max()
        pad = (hi - lo) * pad_frac
        state_ylim.append((lo - pad, hi + pad))

    arr = np.concatenate(ph_data)
    lo, hi = arr.min(), arr.max()
    pad = (hi - lo) * pad_frac
    ph_ylim = (lo - pad, hi + pad)

    return state_ylim, ph_ylim


def new_figure(step, window=20.0):
    current_time = step["t_next"]   # measurements after this time are hidden

    if xlim is not None:
        t_lo, t_hi = xlim
    else:
        t_lo, t_hi = step["t_next"] - window, step["t_next"] + window

    fig, axes = plt.subplots(2, 2, figsize=(10, 7))
    mmask = (t_meas >= t_lo) & (t_meas <= t_hi) & (t_meas <= current_time)

    for i, (r, c) in enumerate(panel_pos):
        ax = axes[r, c]
        ax.set_xlim(t_lo, t_hi)
        ax.set_ylim(*STATE_YLIM[i])
        ax.set_ylabel(state_labels[i], fontsize=14)

    ax_ph = axes[1, 1]
    ax_ph.plot(t_meas[mmask], y_meas[mmask], ".", color=MEASUREMENT_COLOR,
               ms=4, alpha=0.5, label="pH meas. (hist.)")
    ax_ph.set_xlim(t_lo, t_hi)
    ax_ph.set_ylim(*PH_YLIM)
    ax_ph.set_ylabel("pH", fontsize=14)

    # bottom row keeps x-axis labels/ticks; top row is stripped for a shared-x look
    for (r, c) in panel_pos + [(1, 1)]:
        ax = axes[r, c]
        if r == 0:
            ax.set_xlabel("")
            ax.tick_params(axis="x", which="both", bottom=False, labelbottom=False)
        else:
            ax.set_xlabel("time [min]")

    return fig, axes


def jitter(n, scale=0.6, seed=1):
    return (np.random.default_rng(seed).random(n) - 0.5) * scale


def size_from_weight(w, s_min=2.0, s_max=60.0):
    w_lo, w_hi = w.min(), w.max()
    if w_hi - w_lo < 1e-15:
        return np.full_like(w, 0.5 * (s_min + s_max))
    return s_min + (s_max - s_min) * (w - w_lo) / (w_hi - w_lo)
```

### Step 1 — Sampling (static)

```python
def plot_sampling(step, fname):
    fig, axes = new_figure(step)
    t_prev = step["t_prev"]
    x_hat, P = step["x_hat_prior"], step["P_prior"]
    std = np.sqrt(np.diag(P))
    X0 = step["X0"]
    jit = jitter(X0.shape[0])

    for i, (r, c) in enumerate(panel_pos):
        ax = axes[r, c]
        ax.errorbar([t_prev + CI_X_OFFSET], [x_hat[i]], yerr=[1.96 * std[i]], fmt="o",
                    color=CI_COLOR, capsize=4, elinewidth=2, markersize=5, zorder=6,
                    label="95% CI (prior)")
        ax.scatter(t_prev + jit, X0[:, i], color=PARTICLE_COLOR, s=12, alpha=0.6,
                   zorder=5, label="particles")

    axes[0, 0].legend(fontsize=7, loc="upper left")
    fig.suptitle(f"Step 1 -- Sampling  (t = {t_prev:.0f} min)")
    fig.tight_layout()
    fig.savefig(fname, dpi=150)
    plt.close(fig)
```

### Step 2 — Propagate (animated)

```python
def make_propagate_video(step, fname, fps=10):
    fig, axes = new_figure(step)
    traj = step["traj"]                     # (n_sub+1, N, 3)
    n_sub = traj.shape[0] - 1
    N = traj.shape[1]
    t_sub = np.linspace(step["t_prev"], step["t_next"], n_sub + 1)

    scat, lines = [], []
    for i, (r, c) in enumerate(panel_pos):
        ax = axes[r, c]
        sc = ax.scatter(np.full(N, t_sub[0]), traj[0, :, i], color=PARTICLE_COLOR, s=14, zorder=5)
        scat.append(sc)
        ln = [ax.plot([], [], color="gray", lw=0.6, alpha=0.5, zorder=1)[0] for _ in range(N)]
        lines.append(ln)

    fig.suptitle("Step 2 -- Propagate")
    fig.tight_layout()

    def update(frame):
        artists = []
        for i in range(3):
            scat[i].set_offsets(np.column_stack([np.full(N, t_sub[frame]), traj[frame, :, i]]))
            artists.append(scat[i])
            for j in range(N):
                lines[i][j].set_data(t_sub[:frame + 1], traj[:frame + 1, j, i])
            artists.extend(lines[i])
        return artists

    ani = animation.FuncAnimation(fig, update, frames=n_sub + 1, blit=False)
    ani.save(fname, writer=animation.FFMpegWriter(fps=fps))   # plays once, no loop
    plt.close(fig)
```

### Step 3 — Observation model (static)

```python
def plot_observation(step, fname):
    fig, axes = new_figure(step)
    t_next = step["t_next"]
    Xk, y_pred = step["Xk"], step["y_pred"]
    jit = jitter(Xk.shape[0])

    for i, (r, c) in enumerate(panel_pos):
        axes[r, c].scatter(t_next + jit, Xk[:, i], color=PARTICLE_COLOR, s=12, alpha=0.6,
                            zorder=5, label="particles")

    axes[1, 1].scatter(t_next + jit, y_pred, color=PARTICLE_COLOR, edgecolor="k", linewidth=0.3,
                        s=16, alpha=0.85, zorder=5, label="predicted pH per particle")
    axes[1, 1].legend(fontsize=7, loc="upper left")

    fig.suptitle(f"Step 3 -- Observation model  (t = {t_next:.0f} min)")
    fig.tight_layout()
    fig.savefig(fname, dpi=150)
    plt.close(fig)
```

### Step 4 — Measure & weight (animated)

```python
def make_measure_video(step, fname, fps=10, n_hold=5, n_resize=30):
    fig, axes = new_figure(step)
    t_next = step["t_next"]
    Xk, y_pred, w = step["Xk"], step["y_pred"], step["w"]
    y_meas_k = step["y_meas"]
    jit = jitter(Xk.shape[0])

    s_min, s_max = 1.0, 200.0
    s0 = 16                               # neutral starting size: some particles
                                          # will grow, others will shrink from here
    s_final = size_from_weight(w, s_min, s_max)

    state_scat = []
    for i, (r, c) in enumerate(panel_pos):
        sc = axes[r, c].scatter(t_next + jit, Xk[:, i], color=PARTICLE_COLOR,
                                 s=np.full(len(w), s0), alpha=0.7, zorder=5)
        state_scat.append(sc)
    ph_scat = axes[1, 1].scatter(t_next + jit, y_pred, color=PARTICLE_COLOR, edgecolor="k",
                                  linewidth=0.3, s=np.full(len(w), s0), alpha=0.9, zorder=5)

    sigma = np.sqrt(R[0, 0])
    axes[1, 1].errorbar([t_next + CI_X_OFFSET], [y_meas_k], yerr=[1.96 * sigma], fmt="o",
                         color=MEASUREMENT_COLOR, capsize=4, elinewidth=2, markersize=6,
                         zorder=6, label="measurement (95% CI)")
    axes[1, 1].legend(fontsize=7, loc="upper left")
    fig.suptitle("Step 4 -- Measure & weight")
    fig.tight_layout()

    n_frames = n_hold + n_resize

    def update(frame):
        if frame < n_hold:
            s = np.full(len(w), s0)
        else:
            a = min((frame - n_hold + 1) / n_resize, 1.0)
            s = s0 + (s_final - s0) * a
        for sc in state_scat:
            sc.set_sizes(s)
        ph_scat.set_sizes(s)
        return state_scat + [ph_scat]

    ani = animation.FuncAnimation(fig, update, frames=n_frames, blit=False)
    ani.save(fname, writer=animation.FFMpegWriter(fps=fps))
    plt.close(fig)
```

### Step 5 — Update (animated)

```python
def make_update_video(step, fname, fps=10, n_frames=20):
    fig, axes = new_figure(step)
    t_next = step["t_next"]
    Xk, y_pred, w = step["Xk"], step["y_pred"], step["w"]
    x_hat_post, P_post = step["x_hat_post"], step["P_post"]
    std_post = np.sqrt(np.diag(P_post))
    jit = jitter(Xk.shape[0])

    s_min, s_max = 1.0, 200.0
    s_final = size_from_weight(w, s_min, s_max)

    state_scat = []
    for i, (r, c) in enumerate(panel_pos):
        sc = axes[r, c].scatter(t_next + jit, Xk[:, i], color=PARTICLE_COLOR, s=s_final,
                                 alpha=0.8, zorder=5)
        state_scat.append(sc)
    ph_scat = axes[1, 1].scatter(t_next + jit, y_pred, color=PARTICLE_COLOR, edgecolor="k",
                                  linewidth=0.3, s=s_final, alpha=0.9, zorder=5)

    ci_artists = []
    for i, (r, c) in enumerate(panel_pos):
        eb = axes[r, c].errorbar([t_next + CI_X_OFFSET], [x_hat_post[i]], yerr=[1.96 * std_post[i]],
                                  fmt="o", color=CI_COLOR, capsize=4, elinewidth=2, markersize=5,
                                  zorder=6, alpha=0)
        ci_artists.append(eb)

    fig.suptitle("Step 5 -- Update")
    fig.tight_layout()

    def set_errorbar_alpha(eb, alpha):
        eb[0].set_alpha(alpha)
        for coll in eb[1]:
            coll.set_alpha(alpha)
        for coll in eb[2]:
            coll.set_alpha(alpha)

    def update(frame):
        a = frame / (n_frames - 1)
        fade, grow = max(0.0, 1.0 - a), min(1.0, a)
        for sc in state_scat:
            sc.set_alpha(fade * 0.8)
        ph_scat.set_alpha(fade * 0.9)
        for eb in ci_artists:
            set_errorbar_alpha(eb, grow)
        return state_scat + [ph_scat]

    ani = animation.FuncAnimation(fig, update, frames=n_frames, blit=False)
    ani.save(fname, writer=animation.FFMpegWriter(fps=fps))
    plt.close(fig)
```


## Generate the figure set for the chosen step

`step_idx` (and, if set, `xlim`) are defined once in the Visualization setup
cell above. Filenames include `step_idx` so multiple steps can be rendered
and kept side by side without overwriting each other. The 3 videos are MP4s
(play once, no loop) so they behave well when pasted into PowerPoint.


```python
OUT = "./figures"
state_labels = [r"$C_{Cu}^{(aq)}$", r"$C_H$", r"$C_{Cu}^{(s)}$"]
panel_pos = [(0, 0), (0, 1), (1, 0)]   # state panels; (1,1) is reserved for pH


# --- style constants, shared across all 5 figures ---
PARTICLE_COLOR = "darkblue"
MEASUREMENT_COLOR = "darkgoldenrod"     # "dark yellow"
CI_COLOR = "black"
CI_X_OFFSET = -0.8                      # minutes; shifts CI bars left of the markers

# --- step_idx and (optional) x-axis limits, set once here ---
step_idx = 16            # <-- choose which transition to visualize (0 = first)
xlim = None              # <-- e.g. (30, 90) to force fixed x-axis limits on every
                         #     panel; leave as None to auto-window around step_idx

step = history[step_idx]   # used by the single-step section below

STATE_YLIM, PH_YLIM = compute_axis_limits([step], R)   # scoped to the selected step_idx


plot_sampling(step, f"{OUT}/pf_step1_sampling_{step_idx}.png")
make_propagate_video(step, f"{OUT}/pf_step2_propagate_{step_idx}.mp4")
plot_observation(step, f"{OUT}/pf_step3_observation_{step_idx}.png")
make_measure_video(step, f"{OUT}/pf_step4_measure_{step_idx}.mp4")
make_update_video(step, f"{OUT}/pf_step5_update_{step_idx}.mp4")

print(f"Generated figures for step_idx={step_idx}  (t = {step['t_prev']:.0f} -> {step['t_next']:.0f} min)")
```


## Combined animation across multiple steps

The two static frames (Sampling, Observation model) become short *held*
clips so every stage can be concatenated into one continuous video. Set
`step_idx_start` / `step_idx_finish` below to choose the range. This also
re-generates the individual per-step files (same as the single-step section
above) for every step in the range, so you still have each stage available
on its own for pasting into a slide, in addition to the combined video.


```python
def make_sampling_video(step, fname, hold_seconds=1.5, fps=10):
    """Video version of Step 1 (Sampling): a still frame held for
    `hold_seconds`, so it can be concatenated with the animated steps."""
    fig, axes = new_figure(step)
    t_prev = step["t_prev"]
    x_hat, P = step["x_hat_prior"], step["P_prior"]
    std = np.sqrt(np.diag(P))
    X0 = step["X0"]
    jit = jitter(X0.shape[0])

    for i, (r, c) in enumerate(panel_pos):
        ax = axes[r, c]
        ax.errorbar([t_prev + CI_X_OFFSET], [x_hat[i]], yerr=[1.96 * std[i]], fmt="o",
                    color=CI_COLOR, capsize=4, elinewidth=2, markersize=5, zorder=6,
                    label="95% CI (prior)")
        ax.scatter(t_prev + jit, X0[:, i], color=PARTICLE_COLOR, s=12, alpha=0.6,
                   zorder=5, label="particles")

    axes[0, 0].legend(fontsize=7, loc="upper left")
    fig.suptitle(f"Step 1 -- Sampling  (t = {t_prev:.0f} min)")
    fig.tight_layout()

    n_frames = max(1, int(hold_seconds * fps))
    ani = animation.FuncAnimation(fig, lambda frame: [], frames=n_frames, blit=False)
    ani.save(fname, writer=animation.FFMpegWriter(fps=fps))
    plt.close(fig)


def make_observation_video(step, fname, hold_seconds=1.5, fps=10):
    """Video version of Step 3 (Observation model): a still frame held for
    `hold_seconds`, so it can be concatenated with the animated steps."""
    fig, axes = new_figure(step)
    t_next = step["t_next"]
    Xk, y_pred = step["Xk"], step["y_pred"]
    jit = jitter(Xk.shape[0])

    for i, (r, c) in enumerate(panel_pos):
        axes[r, c].scatter(t_next + jit, Xk[:, i], color=PARTICLE_COLOR, s=12, alpha=0.6,
                            zorder=5, label="particles")

    axes[1, 1].scatter(t_next + jit, y_pred, color=PARTICLE_COLOR, edgecolor="k", linewidth=0.3,
                        s=16, alpha=0.85, zorder=5, label="predicted pH per particle")
    axes[1, 1].legend(fontsize=7, loc="upper left")

    fig.suptitle(f"Step 3 -- Observation model  (t = {t_next:.0f} min)")
    fig.tight_layout()

    n_frames = max(1, int(hold_seconds * fps))
    ani = animation.FuncAnimation(fig, lambda frame: [], frames=n_frames, blit=False)
    ani.save(fname, writer=animation.FFMpegWriter(fps=fps))
    plt.close(fig)


def concat_videos(video_paths, output_path):
    """Losslessly concatenate a list of MP4s (same codec/resolution/fps)
    into one file via ffmpeg's concat demuxer."""
    list_file = os.path.join(os.path.dirname(output_path) or ".", "_concat_list.txt")
    with open(list_file, "w") as fh:
        for vp in video_paths:
            fh.write(f"file '{os.path.abspath(vp)}'\n")
    cmd = ["ffmpeg", "-y", "-f", "concat", "-safe", "0", "-i", list_file,
           "-c", "copy", output_path]
    subprocess.run(cmd, check=True, capture_output=True)
    os.remove(list_file)
```

```python
step_idx_start = 15         # <-- first transition to include
step_idx_finish = 25        # <-- last transition to include (inclusive)
hold_seconds = 1.5          # how long each static stage (1, 3) is held in the video
fps = 10                    # shared frame rate across all clips (needed for concatenation)

_margin = 4.0   # minutes of padding on either side
xlim = (history[step_idx_start]["t_prev"] - _margin,
        history[step_idx_finish]["t_next"] + _margin)

STATE_YLIM, PH_YLIM = compute_axis_limits(history[step_idx_start:step_idx_finish + 1], R)

tmp_dir = f"{OUT}/_tmp_stitch"
os.makedirs(tmp_dir, exist_ok=True)

clip_paths = []
for idx in range(step_idx_start, step_idx_finish + 1):
    s = history[idx]

    plot_sampling(s, f"{OUT}/pf_step1_sampling_{idx}.png")
    make_propagate_video(s, f"{OUT}/pf_step2_propagate_{idx}.mp4", fps=fps)
    plot_observation(s, f"{OUT}/pf_step3_observation_{idx}.png")
    make_measure_video(s, f"{OUT}/pf_step4_measure_{idx}.mp4", fps=fps)
    make_update_video(s, f"{OUT}/pf_step5_update_{idx}.mp4", fps=fps)

    # short "held" video versions of the two static stages, only needed for stitching
    samp_vid = f"{tmp_dir}/step1_{idx}.mp4"
    obs_vid = f"{tmp_dir}/step3_{idx}.mp4"
    make_sampling_video(s, samp_vid, hold_seconds=hold_seconds, fps=fps)
    make_observation_video(s, obs_vid, hold_seconds=hold_seconds, fps=fps)

    clip_paths += [
        samp_vid,
        f"{OUT}/pf_step2_propagate_{idx}.mp4",
        obs_vid,
        f"{OUT}/pf_step4_measure_{idx}.mp4",
        f"{OUT}/pf_step5_update_{idx}.mp4",
    ]

combined_path = f"{OUT}/pf_combined_{step_idx_start}_to_{step_idx_finish}.mp4"
concat_videos(clip_paths, combined_path)
shutil.rmtree(tmp_dir)

print(f"Combined animation ({len(clip_paths)} clips) saved to {combined_path}")
display(Video(combined_path, embed=True))
```
